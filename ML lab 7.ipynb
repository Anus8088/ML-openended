```python
# I am using student_por because it has more data than student_mat, I am not merging these to datasets because 383 rows are common in both datasets
data = df_por
data.shape
```

```python
data.head()
```

```python
df_encoded = pd.get_dummies(data, drop_first=True, dtype=int)
```

## PART 1: LINEAR REGRESSION 

```python
from sklearn.model_selection import train_test_split
X_reg = df_encoded.drop('G3', axis=1)
y_reg = df_encoded['G3']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42

)
```

```python
# Scaling features
from sklearn.preprocessing import StandardScaler
scaler_reg = StandardScaler()
X_train_r_scaled = scaler_reg.fit_transform(X_train_r)
X_test_r_scaled = scaler_reg.transform(X_test_r)
```

```python
# Model Training
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train_r_scaled, y_train_r)
```

```python
from sklearn.metrics import mean_squared_error
y_pred_r = lr.predict(X_test_r_scaled)

print("--- REGRESSION RESULT ---")
print("Linear Regression MSE:", mean_squared_error(y_test_r, y_pred_r))
```

## PART 2: CLASSIFICATION

```python
# Target Variable: 1 if G3 >= 10 (Pass), else 0 (Fail)
df_clf = df_encoded.copy()
y_clf = (df_clf['G3'] >= 10).astype(int)
X_clf = df_clf.drop('G3', axis=1)
```

```python
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)
```

```python
# Scaling features
scaler_clf = StandardScaler()
X_train_c_scaled = scaler_clf.fit_transform(X_train_c)
X_test_c_scaled = scaler_clf.transform(X_test_c)
```

### 1. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_c_scaled, y_train_c)
```

```python
pred_log = log_reg.predict(X_test_c_scaled)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
print("Logistic Regression Metrics:")
print("  Accuracy :", accuracy_score(y_test_c, pred_log))
print("  Precision:", precision_score(y_test_c, pred_log))
print("  Recall   :", recall_score(y_test_c, pred_log))
print("  Confusion Matrix:\n", confusion_matrix(y_test_c, pred_log))
```

# 2. Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train_c, y_train_c)
```

```python
pred_dt = dt.predict(X_test_c)
```

```python
print("\nDecision Tree Metrics:")
print("  Accuracy :", accuracy_score(y_test_c, pred_dt))
print("  Precision:", precision_score(y_test_c, pred_dt))
print("  Recall   :", recall_score(y_test_c, pred_dt))
print("  Confusion Matrix:\n", confusion_matrix(y_test_c, pred_dt))
```
