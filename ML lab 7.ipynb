```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, mean_squared_error

# Data load
data = pd.read_csv('student-por.csv', sep=';')

# --- PART 1: LINEAR REGRESSION (Final marks prediction) ---
scale_cols = data.select_dtypes(include=['int64','float64']).columns.tolist()
scale_cols.remove('G3')

scaler = StandardScaler()
data[scale_cols] = scaler.fit_transform(data[scale_cols])

X_reg = data.drop('G3', axis=1)
y_reg = data['G3'] # 0-20 marks

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

lr = LinearRegression()
lr.fit(X_train_r, y_train_r)
y_pred_r = lr.predict(X_test_r)
print("Linear Regression MSE:", mean_squared_error(y_test_r, y_pred_r))

# --- PART 2: CLASSIFICATION (Pass/Fail) ---
# Create binary target
data_class = data.copy()
data_class['G3_binary'] = data_class['G3'].apply(lambda x: 1 if x >= 10 else 0) # 1=Pass, 0=Fail

X_clf = data_class.drop(['G3', 'G3_binary'], axis=1)
y_clf = data_class['G3_binary']

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)

# Logistic Regression
log_reg = LogisticRegression()
log_reg.fit(X_train_c, y_train_c)
pred_log = log_reg.predict(X_test_c)

# Decision Tree
dt = DecisionTreeClassifier()
dt.fit(X_train_c, y_train_c)
pred_dt = dt.predict(X_test_c)

# Compare
print("\nLogistic Regression:")
print("Accuracy:", accuracy_score(y_test_c, pred_log))
print("Precision:", precision_score(y_test_c, pred_log))
print("Recall:", recall_score(y_test_c, pred_log))

print("\nDecision Tree:")
print("Accuracy:", accuracy_score(y_test_c, pred_dt))
print("Precision:", precision_score(y_test_c, pred_dt))
print("Recall:", recall_score(y_test_c, pred_dt))
```
